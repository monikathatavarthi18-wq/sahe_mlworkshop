```python
#data Visualization__.matplotlib,seaborn,plotly
import matplotlib.pyplot as plt
```

```python
counts,edges,bars=plt.hist(data['Age'],bins=25,color='green')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title('Age Distrubution')
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
index=data['Survived'].value_counts().keys()
print(index)
labels=data['Survived'].value_counts().values
print(labels)
```

```python
#pie chart
plt.pie(x=labels,
       labels=['Dead','Alive'],
       colors=['red','green'],
       autopct="%.2f%%",
           shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pie chart
plt.pie(x=[577,314],
       labels=['Male','Female'],
       colors=['blue','pink'],
       autopct="%.2f%%",
           shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pairwise relationship _>scatterlot
plt.scatter(x=data['Age'],
            y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```

```python
from plotly.offline import init_notebook_mode,iplot
import plotly.graph_objects as go
init_notebook_mode(connected = True)
import plotly.graph_objects as go
```

```python
#Let's visualize histgram
trace =go.Histogram(x = data['Age'],
                    nbinsx=20,
                    marker={'color':'blue'})
#layout
layout=go.Layout(title="Age Distribution")
fig=go.Figure(data=trace,layout=layout)
iplot(fig)
```

```python
trace=go.Pie(labels=['Dead','Alive'],
             values=[549,342])
layout=go.Layout(title='Survived Distrubition')
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
trace=go.Box(x=data['Pclass'],
             y=data['Age'],
             marker={'color' :'Purple'})
layout=go.Layout(title="age vs pclass distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
trace=go.Scatter(x=data['Age'],y=data['Fare'],
                 mode='markers',
                 marker={'color':'red'})
layout=go.Layout(title= "Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
data.isnull().sum()
```

```python
data['Embarked']=data['Embarked'].fillna('S',inplace=True)
```
