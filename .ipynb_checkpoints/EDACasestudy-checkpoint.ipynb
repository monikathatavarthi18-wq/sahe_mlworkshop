```python
#data Visualization__.matplotlib,seaborn,plotly
import matplotlib.pyplot as plt
```

```python
counts,edges,bars=plt.hist(data['Age'],bins=25,color='green')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title('Age Distrubution')
plt.show()
```

```python
#bar plot representation
data['Sex'].value_counts().plot(kind='bar')
```

```python
#bar plot representation
data['Pclass'].value_counts().plot(kind='barh',color='violet')
```

```python
plt.bar(data['Survived'],data['Sex'])
plt.xlabel("Survived Rate")
plt.ylabel("Gender")
plt.title("Gender Survived Distrubution")
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
index=data['Survived'].value_counts().keys()
print(index)
labels=data['Survived'].value_counts().values
print(labels)
```

```python
#pie chart
plt.pie(x=labels,
       labels=['Dead','Alive'],
       colors=['red','green'],
       autopct="%.2f%%",
           shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pie chart
plt.pie(x=[577,314],
       labels=['Male','Female'],
       colors=['blue','pink'],
       autopct="%.2f%%",
           shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pairwise relationship _>scatterlot
plt.scatter(x=data['Age'],
            y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
